```python
import random
class stonepaperscissor:
    def __init__(self,chance):
        self.bonus=0
        self.chance = chance
        print("0-stone,1-paper,2-scissor,3- quit")
    def game(self):
        self.random_number = random.randint(0,2)
        if self.chance==self.random_number:
            print("bad choice")
        elif self.chance==1 and self.random_number==0:
            self.bonus= self.bonus+1
            print("you win")
        elif self.chance == 2 and self.random_number == 1:
            self.bonus += 1
            print("You win!")
        elif self.chance == 0 and self.random_number == 2:
            self.bonus += 1
            print("You win")
        else:
            print("You lose")
        return self.bonus,self.random_number
        self.play()
        
    def play(self):
        b=int(input("enter choice:"))
        if b==3:
            print("game over")
        elif b > 3:
            print("Invalid choice. Please enter a number between 0 and 3.")
        else:
            self.chance = b 
            bonus, computer_choice = self.game()
            print(f"Your current bonus score : {bonus}")
            print(f" computer_choice : {computer_choice}")
            self.play()
        
chance= 0
result= stonepaperscissor(chance)
result.play()
```
